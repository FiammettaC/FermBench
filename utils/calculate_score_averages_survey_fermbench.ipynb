```python
def calculate_averages(df, num_columns):
    # --- Create a sample DataFrame with categorical string column names ---


    # Define a set of 8 base categorical names that repeat
    base_categories = ["Fact", "Compr", "Coher", "Concis", "Readab", "Helpful", "Insight", "Qual"]

    # Generate column names like 'Gender_0', 'Age_1', ..., 'Employment_7', 'Gender_8', ...
    column_names = []
    for i in range(num_columns):
        base_name = base_categories[i % 8] # This gets the repeating category name
        sequence_number = i // 8
        column_names.append(f"{base_name}_{sequence_number}") # Appends the index for uniqueness if needed


    df = df.set_axis(column_names, axis=1)
    columns_for_averaging = {
    base: {group_idx: [] for group_idx in range(5)} # 5 groups per base_category (0-4)
    for base in base_categories}

    # First, populate the 'columns_for_averaging' dictionary
    for col_name in df.columns:
        parts = col_name.split('_')
        base_name = parts[0]
        sequence_number = int(parts[1])
        average_group_index = sequence_number % 5 # This is the new grouping logic!
        columns_for_averaging[base_name][average_group_index].append(col_name)

    # Now, calculate the averages
    final_averages = pd.DataFrame(index=range(5), columns=base_categories)

    for base_name in base_categories:
        print(f"\n--- {base_name} Averages ---")
        for average_group_index in range(5):
            cols_to_average = columns_for_averaging[base_name][average_group_index]
            if cols_to_average:
                # Calculate the average of all values within these specific columns
                # .mean().mean() takes the average of each column, then averages those averages
                avg_value = df[cols_to_average].mean().mean()
                final_averages.loc[average_group_index, base_name] = avg_value
                print(f"  Group {average_group_index} (suffixes ending in {average_group_index % 5}): {avg_value:.4f}")
            else:
                final_averages.loc[average_group_index, base_name] = np.nan # Or 0.0
                print(f"  Group {average_group_index} (suffixes ending in {average_group_index % 5}): No columns found.")

    print("\n--- Summary DataFrame of Averages ---")
    print(final_averages.round(4))
    return final_averages
final_averages_exp1 = calculate_averages(df, 160)
```

```python
import numpy as np
from itertools import zip_longest

def krippendorff_alpha_ordinal(data, k=5):
    """
    Computes Krippendorff's alpha for ordinal data.
    Parameters:
        data: 2D array (rows = items, columns = raters), may contain np.nan
        k: number of ordinal categories (e.g. 5-point scale → k=5)
    Returns:
        Krippendorff's alpha value
    """
    # Distance matrix for ordinal data
    distance_matrix = np.zeros((k, k))
    for i in range(k):
        for j in range(k):
            distance_matrix[i, j] = ((i - j) ** 2) / ((k - 1) ** 2)

    # Create value coincidence matrix
    coincidence = np.zeros((k, k))
    for row in data:
        row = row[~np.isnan(row)]
        for i in range(len(row)):
            for j in range(i + 1, len(row)):
                a, b = int(row[i]) - 1, int(row[j]) - 1
                coincidence[a, b] += 1
                coincidence[b, a] += 1

    # Total weight
    total_weight = np.sum(coincidence)

    # Observed disagreement (Do)
    Do = 0.0
    for i in range(k):
        for j in range(k):
            Do += distance_matrix[i, j] * coincidence[i, j]
    Do /= total_weight

    # Expected disagreement (De)
    marginals = np.sum(coincidence, axis=0)
    De = 0.0
    for i in range(k):
        for j in range(k):
            De += distance_matrix[i, j] * marginals[i] * marginals[j]
    De /= (total_weight * (total_weight - 1))

    alpha = 1 - Do / De if De != 0 else 1.0
    return alpha

# Example: paste your data here as a list of numpy arrays
data = data_list

# Step 1: Pad shorter arrays with NaN so they align
max_len = max(len(arr) for arr in data)
padded_data = np.array(list(zip_longest(*data, fillvalue=np.nan)))

# Step 2: Compute Krippendorff’s alpha for ordinal data (scale 1-5)
alpha = krippendorff_alpha_ordinal(padded_data, k=5)
print(f"Krippendorff's alpha (ordinal): {alpha:.4f}")
```

```python
num_original_rows = 13
num_target_columns = 40
num_blocks_to_stack = df_original.shape[1] // num_target_columns # 160 // 40 = 4

if df_original.shape[1] % num_target_columns != 0:
    print(f"Warning: Original columns ({df_original.shape[1]}) not perfectly divisible by target columns ({num_target_columns}). Trailing columns will be ignored or require special handling.")

reshaped_blocks = []
for i in range(num_blocks_to_stack):
    # Calculate the start and end index for the current block of columns
    start_col_idx = i * num_target_columns
    end_col_idx = (i + 1) * num_target_columns

    # Select the slice of columns for the current block
    current_block_df = df_original.iloc[:, start_col_idx:end_col_idx]

    # Optionally, rename columns to a generic set if you want consistent names across blocks
    # For instance, 'Column_0' to 'Column_39' for the reshaped DataFrame
    current_block_df.columns = [f'Reshaped_Col_{j}' for j in range(num_target_columns)]

    reshaped_blocks.append(current_block_df)

# Concatenate all the reshaped blocks vertically
df_reshaped = pd.concat(reshaped_blocks, ignore_index=True) # ignore_index=True creates a new clean index

print("\nReshaped DataFrame Info:")
print(f"Shape: {df_reshaped.shape}")
print(f"Head of Reshaped DataFrame:\n{df_reshaped.head()}")
print(f"\nTail of Reshaped DataFrame:\n{df_reshaped.tail()}")

# --- Verification ---
# Let's check a specific cell to see how values moved
# Original df_original[0, 40] (first row, 41st column) should now be in df_reshaped[5, 0] (6th row, 1st column)
# Because original row 0's 41st column (index 40) is in the second block (i=1),
# and it becomes the 0th column of that block.
# Since each block adds 'num_original_rows' new rows, the new row index will be 0 + num_original_rows * block_index
# So, for block index 1 (second block), the original first row becomes row 5.
original_val = df_original.iloc[0, 40] # Value at row 0, col index 40
reshaped_val = df_reshaped.iloc[num_original_rows * 1 + 0, 0] # Value at (0 + 5*1), 0 => (5, 0)

print(f"\nVerification: Value at df_original[0, 40] is {original_val}")
print(f"Value at df_reshaped[{num_original_rows * 1 + 0}, 0] is {reshaped_val}")
print(f"Are they equal? {original_val == reshaped_val}")
```

```python
import pandas as pd
from scipy.stats import friedmanchisquare
import scikit_posthocs as sp

# Define your structure
models = ['Gemini', 'DeepSeek', 'le Chat', 'ChatGPT', 'Claude']
metrics = [
    'Factual Accuracy', 'Comprehensiveness', 'Coherence',
    'Concision', 'Readability', 'Helpfulness',
    'Insightfulness', 'Overall Quality'
]

def run_human_stats(df):
    all_results = {}

    for i, metric in enumerate(metrics):
        # Step 1: Slice the 5 columns for the current metric
        # Columns 0-4 = Metric 1, 5-9 = Metric 2, etc.
        start_idx = i * 5
        end_idx = start_idx + 5
        metric_data = df.iloc[:, start_idx:end_idx].copy()
        metric_data.columns = models

        # Step 2: Clean data (Friedman requires no missing values in a row)
        metric_data = metric_data.dropna()

        print(f"\n--- {metric.upper()} (N={len(metric_data)}) ---")

        # Step 3: Global Friedman Test
        # Unpacks the 5 columns as individual arrays
        stat, p = friedmanchisquare(*[metric_data[m] for m in models])
        print(f"Friedman p-value: {p:.4e}")

        # Step 4: Pairwise Nemenyi Test (if Friedman is significant)
        if p < 0.05:
            # posthoc_nemenyi_friedman expects rows as blocks and columns as groups
            posthoc = sp.posthoc_nemenyi_friedman(metric_data)
            all_results[metric] = posthoc
            print("Significant differences found between pairs:")
            print(posthoc)
        else:
            print("No significant difference found by human annotators.")

    return all_results

# Execute:
results = run_human_stats(df_cleaned)
```
